## 4. Konsep HCI dan Visualisasi Data

**Human-Computer Interaction (HCI)** atau Interaksi Manusia dan Komputer mempelajari bagaimana manusia berinteraksi dengan sistem komputer melalui perancangan, implementasi, dan evaluasi antarmuka.

Tiga komponen utama HCI:
1. **Manusia:** pengguna dengan kebutuhan, kemampuan, dan keterbatasan.
2. **Interaksi:** tindakan dan umpan balik antara pengguna dan sistem.
3. **Komputer:** perangkat dan perangkat lunak yang menjalankan sistem.

**Visualisasi data** mengubah data menjadi grafik, diagram, atau bentuk visual lain agar pola, tren, perbandingan, dan anomali lebih mudah dikenali.

## 8. Kebutuhan Sistem

### Kebutuhan fungsional
- Menampilkan jumlah respons survei.
- Menampilkan rata-rata skor kepuasan.
- Membandingkan rata-rata skor antar kategori.
- Menampilkan tren jumlah respons per bulan.
- Memungkinkan pengguna menyaring data berdasarkan periode atau destinasi.

### Kebutuhan nonfungsional
- Label dan navigasi konsisten.
- Grafik memiliki judul, satuan, dan keterangan yang jelas.
- Antarmuka responsif pada ukuran layar berbeda.
- Kontras warna memadai dan informasi tidak bergantung pada warna saja.
- Data diperbarui secara konsisten dan dapat ditelusuri sumbernya.

## 9. Prinsip HCI yang Diterapkan

- **Visibility:** status filter dan ringkasan utama terlihat jelas.
- **Feedback:** sistem memberi tanda ketika filter diterapkan atau data sedang dimuat.
- **Consistency:** format angka, label, warna kategori, dan navigasi digunakan secara konsisten.
- **Affordance:** kontrol terlihat seperti sesuatu yang bisa digunakan, misalnya tombol atau pilihan filter.
- **Constraint:** pilihan input dibatasi pada nilai yang valid.
- **Recognition rather than recall:** pengguna dapat melihat label dan petunjuk tanpa harus mengingat kode kategori.
- **User-Centered Design:** keputusan desain didasarkan pada kebutuhan dan pengujian pengguna.

## 10. Menyiapkan Data Simulasi



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(2026)
months = pd.date_range("2026-01-01", periods=8, freq="MS")
categories = ["Kebersihan", "Kenyamanan", "Fasilitas", "Pelayanan"]

rows = []
base_scores = {
    "Kebersihan": 4.0,
    "Kenyamanan": 4.1,
    "Fasilitas": 3.7,
    "Pelayanan": 4.2
}

for month_idx, month in enumerate(months):
    monthly_responses = int(rng.integers(45, 110))
    for category in categories:
        score = np.clip(
            base_scores[category] + 0.04 * month_idx + rng.normal(0, 0.18),
            1, 5
        )
        rows.append({
            "bulan": month,
            "kategori": category,
            "rata_rata_skor": round(float(score), 2),
            "jumlah_respons": monthly_responses
        })

df = pd.DataFrame(rows)
df.head()
```

## 11. Memeriksa Data



```python
print("Ukuran data:", df.shape)
print("\nTipe data:")
print(df.dtypes)
print("\nNilai kosong:")
print(df.isna().sum())
print("\nDuplikasi:")
print(df.duplicated().sum())
print("\nRentang skor:", df["rata_rata_skor"].min(), "sampai", df["rata_rata_skor"].max())
```

## 12. Visualisasi 1 — Perbandingan Kategori


```python
avg_category = (
    df.groupby("kategori", as_index=False)["rata_rata_skor"]
      .mean()
      .sort_values("rata_rata_skor", ascending=False)
)

plt.figure(figsize=(8, 4.5))
plt.bar(avg_category["kategori"], avg_category["rata_rata_skor"])
plt.ylim(0, 5)
plt.title("Rata-rata Skor Kepuasan per Kategori")
plt.xlabel("Kategori")
plt.ylabel("Skor rata-rata (1–5)")
plt.tight_layout()
plt.show()

avg_category
```

## 13. Visualisasi 2 — Tren dari Waktu ke Waktu



```python
monthly_trend = (
    df.pivot(index="bulan", columns="kategori", values="rata_rata_skor")
      .sort_index()
)

plt.figure(figsize=(9, 5))
for category in monthly_trend.columns:
    plt.plot(monthly_trend.index, monthly_trend[category], marker="o", label=category)

plt.ylim(1, 5)
plt.title("Tren Skor Kepuasan per Bulan")
plt.xlabel("Bulan")
plt.ylabel("Skor rata-rata (1–5)")
plt.legend(title="Kategori")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

**Interpretasi:** tren membantu melihat perubahan dari waktu ke waktu. Jangan menyimpulkan bahwa perubahan disebabkan oleh suatu program tanpa bukti tambahan. Periksa juga jumlah respons dan apakah metode survei tetap konsisten setiap bulan.

## 14. Visualisasi 3 — Jumlah Respons per Bulan

Jumlah respons membantu memberi konteks terhadap skor. Bulan dengan respons lebih sedikit dapat memiliki ketidakpastian yang lebih besar, sehingga skor perlu dibaca dengan hati-hati.

```python
monthly_responses = (
    df.groupby("bulan", as_index=False)["jumlah_respons"]
      .max()
)

plt.figure(figsize=(9, 4))
plt.bar(monthly_responses["bulan"].dt.strftime("%b %Y"), monthly_responses["jumlah_respons"])
plt.title("Jumlah Respons Survei per Bulan")
plt.xlabel("Bulan")
plt.ylabel("Jumlah respons")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## 15. Prinsip Visualisasi Data yang Baik

- Pilih grafik sesuai pertanyaan: batang untuk perbandingan, garis untuk tren, scatterplot untuk hubungan dua variabel numerik.
- Beri judul, label sumbu, satuan, dan keterangan yang diperlukan.
- Hindari efek 3D, dekorasi berlebihan, dan garis bantu yang tidak berguna.
- Gunakan skala yang jujur; jangan memotong sumbu untuk membesar-besarkan perbedaan.
- Gunakan warna secara konsisten dan pastikan informasi tetap bisa dipahami tanpa mengandalkan warna saja.
- Sertakan sumber, periode, dan konteks data agar pengguna tidak salah menafsirkan hasil.

Prinsip data-ink ratio menekankan bahwa elemen visual sebaiknya membantu menyampaikan informasi, bukan sekadar menjadi hiasan.

## 17. Prototipe dan Evaluasi Usability

Proses desain dilakukan berulang melalui analisis, desain, prototipe, dan evaluasi.

- **Low-fidelity:** sketsa tata letak atau wireframe sederhana untuk menguji struktur.
- **High-fidelity:** desain digital yang lebih mendekati tampilan akhir.
- **Evaluasi ahli:** pemeriksaan terhadap prinsip heuristik dan konsistensi antarmuka.
- **Pengujian pengguna:** meminta pengguna menyelesaikan tugas nyata dan mengamati kesulitan yang terjadi.

Contoh tugas uji:
1. Temukan kategori dengan skor rata-rata terendah.
2. Cari tren kepuasan pada dua bulan terakhir.
3. Ubah filter periode.
4. Jelaskan kesimpulan yang dapat diambil dari grafik tanpa membuat klaim sebab-akibat.

## 18. Heuristik Nielsen untuk Evaluasi

Sepuluh prinsip heuristik Nielsen dapat digunakan sebagai daftar periksa:

1. Visibility of system status.
2. Match between system and the real world.
3. User control and freedom.
4. Consistency and standards.
5. Error prevention.
6. Recognition rather than recall.
7. Flexibility and efficiency of use.
8. Aesthetic and minimalist design.
9. Help users recognize, diagnose, and recover from errors.
10. Help and documentation.

Evaluasi perlu mencatat masalah yang ditemukan, dampaknya terhadap pengguna, dan usulan perbaikannya. Prinsip heuristik membantu menemukan masalah, tetapi tidak menggantikan pengujian langsung dengan pengguna.

## 19. Aksesibilitas, Dark Patterns, dan Etika

- Gunakan teks yang mudah dibaca dan kontras warna yang memadai.
- Sediakan label yang jelas untuk tombol dan kontrol.
- Jangan mengandalkan warna saja untuk membedakan kategori.
- Pastikan navigasi dapat dipahami dan kontrol mudah dioperasikan.
- Hindari dark patterns, seperti menyamarkan pilihan, mempersulit pembatalan, atau memanipulasi pengguna agar menyetujui sesuatu.
- Sajikan grafik secara jujur dan jelaskan konteks serta keterbatasan data.
- Lindungi informasi pribadi dalam data survei dan batasi akses sesuai kebutuhan.
