### F. Memahami metrik klasifikasi
- **Accuracy:** proporsi seluruh prediksi yang benar. Dapat menyesatkan jika kelas tidak seimbang.
- **Precision kelas terlambat:** dari semua tagihan yang diprediksi terlambat, berapa proporsi yang benar-benar terlambat.
- **Recall kelas terlambat:** dari semua tagihan yang benar-benar terlambat, berapa proporsi yang berhasil dikenali.
- **F1-score:** rata-rata harmonik precision dan recall.
- **ROC-AUC:** kemampuan model mengurutkan contoh positif di atas negatif pada berbagai ambang; bukan ukuran kalibrasi probabilitas.
- **Confusion matrix:** menunjukkan jumlah benar/salah untuk masing-masing kelas.

Recall tinggi dapat membantu mengurangi tagihan terlambat yang terlewat, tetapi mungkin menambah tindak lanjut yang tidak perlu. Precision tinggi mengurangi alarm palsu, tetapi bisa melewatkan lebih banyak kasus. Pilihan metrik perlu mengikuti biaya operasional dan tujuan bisnis.

## 2. Import Library

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, classification_report, ConfusionMatrixDisplay
)

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

print("Library siap digunakan.")
```

## 3. Data Understanding



```python
# Membuat dataset simulasi (bukan data perusahaan nyata)
n = 1200

jumlah_tagihan = np.random.lognormal(mean=14.0, sigma=0.65, size=n).round(0)
hari_menuju_jatuh_tempo = np.random.randint(1, 31, size=n)
riwayat_telat_sebelumnya = np.random.poisson(lam=1.2, size=n)
jenis_pelanggan = np.random.choice(
    ["Retail", "Bisnis Kecil", "Perusahaan"], size=n, p=[0.45, 0.35, 0.20]
)
metode_pembayaran = np.random.choice(
    ["Transfer", "Kartu", "Virtual Account"], size=n, p=[0.50, 0.20, 0.30]
)

# Membuat probabilitas keterlambatan berdasarkan beberapa faktor
logit = (
    -2.0
    + 0.55 * riwayat_telat_sebelumnya
    + 0.000003 * jumlah_tagihan
    + np.where(jenis_pelanggan == "Retail", 0.25, 0)
    + np.where(metode_pembayaran == "Transfer", 0.20, 0)
)
prob_telat = 1 / (1 + np.exp(-logit))
terlambat = np.random.binomial(1, prob_telat)

df = pd.DataFrame({
    "jumlah_tagihan": jumlah_tagihan,
    "hari_menuju_jatuh_tempo": hari_menuju_jatuh_tempo,
    "riwayat_telat_sebelumnya": riwayat_telat_sebelumnya,
    "jenis_pelanggan": jenis_pelanggan,
    "metode_pembayaran": metode_pembayaran,
    "terlambat": terlambat
})

display(df.head())
print("Ukuran dataset:", df.shape)
print("\nDistribusi target:")
display(df["terlambat"].value_counts().rename(index={0: "Tidak terlambat", 1: "Terlambat"}).to_frame("jumlah"))
```

```python
# Periksa tipe data, nilai kosong, dan duplikasi
print("Informasi dataset:")
df.info()

print("\nJumlah missing value per kolom:")
display(df.isna().sum().to_frame("missing_value"))

print("\nJumlah baris duplikat:", df.duplicated().sum())

print("\nStatistik numerik:")
display(df.describe())
```

## 4. Exploratory Data Analysis (EDA)



```python
# Distribusi keterlambatan
target_counts = df["terlambat"].value_counts().sort_index()
target_counts.index = ["Tidak terlambat (0)", "Terlambat (1)"]
target_counts.plot(kind="bar")
plt.title("Distribusi Status Pembayaran")
plt.ylabel("Jumlah tagihan")
plt.xticks(rotation=0)
plt.show()

# Rata-rata jumlah tagihan berdasarkan status
display(
    df.groupby("terlambat")[["jumlah_tagihan", "riwayat_telat_sebelumnya"]]
      .mean()
      .rename(index={0: "Tidak terlambat", 1: "Terlambat"})
)
```

## 5. Data Preparation



```python
target = "terlambat"
X = df.drop(columns=[target])
y = df[target]

numeric_features = [
    "jumlah_tagihan",
    "hari_menuju_jatuh_tempo",
    "riwayat_telat_sebelumnya"
]
categorical_features = ["jenis_pelanggan", "metode_pembayaran"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=RANDOM_STATE, stratify=y
)

numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

print("Data latih:", X_train.shape)
print("Data uji:", X_test.shape)
```

## 6. Modeling: Membandingkan Beberapa Model

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, class_weight="balanced"),
    "Decision Tree": DecisionTreeClassifier(
        max_depth=5, random_state=RANDOM_STATE, class_weight="balanced"
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=200, random_state=RANDOM_STATE,
        class_weight="balanced", max_depth=8
    )
}

results = []
fitted_models = {}

for name, estimator in models.items():
    pipeline = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", estimator)
    ])
    pipeline.fit(X_train, y_train)
    y_pred = pipeline.predict(X_test)
    y_prob = pipeline.predict_proba(X_test)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1-score": f1_score(y_test, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })
    fitted_models[name] = pipeline

results_df = pd.DataFrame(results).sort_values("Recall", ascending=False)
display(results_df.round(3))
```

## 7. Evaluation



```python
# Pilih model dengan recall tertinggi pada hasil uji sebagai contoh aturan pemilihan
best_model_name = results_df.iloc[0]["Model"]
best_model = fitted_models[best_model_name]
best_pred = best_model.predict(X_test)

print("Model terpilih berdasarkan recall tertinggi:", best_model_name)
print("\nClassification report:")
print(classification_report(
    y_test, best_pred,
    target_names=["Tidak terlambat", "Terlambat"],
    zero_division=0
))

ConfusionMatrixDisplay.from_predictions(
    y_test, best_pred,
    display_labels=["Tidak terlambat", "Terlambat"],
    values_format="d"
)
plt.title(f"Confusion Matrix - {best_model_name}")
plt.show()
```

## 8. Contoh Prediksi Tagihan Baru



```python
tagihan_baru = pd.DataFrame([{
    "jumlah_tagihan": 1800000,
    "hari_menuju_jatuh_tempo": 5,
    "riwayat_telat_sebelumnya": 3,
    "jenis_pelanggan": "Bisnis Kecil",
    "metode_pembayaran": "Transfer"
}])

prediksi = int(best_model.predict(tagihan_baru)[0])
probabilitas = float(best_model.predict_proba(tagihan_baru)[0, 1])

print("Prediksi:", "Berpotensi terlambat" if prediksi == 1 else "Tidak terindikasi terlambat")
print(f"Probabilitas menurut model: {probabilitas:.1%}")
print("Gunakan hasil ini sebagai alat bantu, bukan satu-satunya dasar keputusan.")
```
